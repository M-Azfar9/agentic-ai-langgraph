```python
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

load_dotenv()  
```

```python
prompt = """
Implement a thread-safe LFU (Least Frequently Used) Cache in Python.

Requirements:
- O(1) average time complexity for get() and put().
- Do NOT use external libraries.
- Use object-oriented design.
- Include type hints.
- Add clear comments.
- Explain the data structures used.
- Analyze time and space complexity.
- Provide at least 8 unit tests covering edge cases.
"""
```

```python
mistral = ChatOpenRouter(
    model="nvidia/nemotron-3-ultra-550b-a55b:free",
)

for chunk in mistral.stream(prompt):
    print(chunk.content, end="", flush=True)

print()  # New line after streaming completes
```

```python
glm = ChatOpenRouter(
    model="z-ai/glm-5.2",
)

glm_response = glm.invoke(prompt)

print(glm_response.content)
```

```python
gpt = ChatOpenRouter(
    model="qwen/qwen3-coder:free",
)

gpt_response = gpt.invoke(prompt)

print(gpt_response.content)
```

```python
nemotron = ChatOpenRouter(
    model="nvidia/nemotron-3-ultra-550b-a55b:free",
)
response = ""
for chunk in nemotron.stream("Write enhance version of signin and signup UI code in Flutter, use Claymorphism UI design style"):
    if chunk.content:
        print(chunk.content, end="", flush=True)
        response += chunk.content
```
