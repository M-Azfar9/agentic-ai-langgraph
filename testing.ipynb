```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
question = "Sally has 3 brothers. Each of her brothers has 2 sisters. How many sisters does Sally have?"
```

```python
gemini = ChatGoogleGenerativeAI(model='gemini-2.5-flash')
```

```python
response = gemini.invoke(question)
print(response.content)
```

```python
# 2. Initialize the base endpoint with the 'conversational' task
llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-R1",
    task="conversational",  # Must match the provider's supported task
    max_new_tokens=1024,
    temperature=0.7,
    timeout=300
)

# 3. WRAP the llm in ChatHuggingFace
# This is the critical step to avoid the ValueError
chat_model = ChatHuggingFace(llm=llm)
```

```python
# Stream with proper formatting
for chunk in chat_model.stream(question):
    # end="" keeps it on one line, flush=True ensures it updates immediately
    print(chunk.content, end="", flush=True)
```
